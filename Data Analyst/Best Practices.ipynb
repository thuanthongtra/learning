# DAX


## Measures
- Keep DAX measures **_grouped together_**
  - Measure are free to move to any table
  - In contrast with **_calculated columns_**, which **_belong to a specific table_**
- Format and comment DAX
  - Indentation for increasing understanding
  - Add comments after `//`

### Use Variables
- Stores result of expression as a named variable
- Advantages:
  - Improve performance
  - Improve readability
  - Simplify debugging
  - Reduce complexity
- Syntax: `VAR <name> = <expression> RETURN`

# Modeling

## Removing Bi-directional

### Using Filter Measure
**The purpose:** to choose values from 1 dimension tables will filter out values of other dimension table

**Example:** Fact table **_MyFactTable_** is filterred by 2 dimensional tables: **_Year_** + **_Geography_** (single direction from both dim tables)
1. Create new measure in Fact table: **_Slicer_MyFactTable_**
    - `Slicer_MyFactTable = INT(NOT ISEMPTY('MyFactTable'))`
    - Returns 1 if atleast one value in the fact table
    - Returns 0 if no values in the fact table
2. Add a **_visual filter_** to the slicer and set where `Slicer_MyFactTable = 1`

### CROSSFILTER()
**The purpose:** to calculate from 1 dimensional table with associated number of another dimensional table

**Example:** Get the count of distinct **_ProductCategory_** associated to a **_AccountName_** with the modelling:

![](./images/Datacamp/CROSSFILTER_Removing_Bi_Direction_Example_1.png)
- We can easily achieve this by changing the **_direction of (2) = both_**. But it might come with performance issues.
- Using normal DISTINCTCOUNT we cannot get the correct result
  - `Product_Category_Count = DISTINCTCOUNT(Dim_ProductCategory[ProductCategoryDescription])`
  - The result: **_incorrect_**

    ![](./images/Datacamp/CROSSFILTER_Removing_Bi_Direction_Example_2.png)

- Adding CROSSFILTER to overwrite the current direction in modelling
  ```Product_Category_Count = 
          CALCULATE(
              DISTINCTCOUNT(Dim_ProductCategory[ProductCategoryDescription]),
              CROSSFILTER(Dim_ProductCategory[ProductCategoryKey], Fact_StrategyPlan[ProductCategoryKey], BOTH)
          )
  ```

  - The result: **_correct_**
  
    ![](./images/Datacamp/CROSSFILTER_Removing_Bi_Direction_Example_3.png)
