# Modeling

## Clean Data
- Free from missing values
- Typos/data entry errors are corrected
- No duplicates
- Irrelevant data is excluded
- Outliers are dealt with
- Correct data types for all columns
- Colum & table names are short and descriptive

## Performance Tuning Advice
**DAX Query slowness**
- Tune DAX operations
- Improve data loading performance

**Visual Display Slowness**
- Use less complicated visuals
- Show less information on the screen

**Other Slowness**
- Reduce number of visuals on the page

## Performance Problems
- Data import
- Querying the database with DirectQuery
- Displaying visuals
- Calculated versus computed columns
- Inefficient relationships
    - M:M relationships
    - Bi-directional cross-filtering

### Optimizing Data Import
- Remove unneccessary rows and columns
- Choose correct data types
  - Numeric data takes less space
  - Casting and aggregating data is slower
- Group and summarize data
  - Store less data on disk
  - Get to aggregate results faster
- Limit parallel queries
- Relationsal database advice
  - Write efficient SQL queries
  - Use appropriate indexes
  - Get the right columns and rows

### Calculated vs. Computed columns
|Calcualted columns | Computed columns|
|-|-|
|DAX| Power Query (M)|
|**Fast** for _simple_ calculatations| **Fast** for _simple_ calculatations|
|**Slow** for _complex_ calculatations|**Fast** for _complex_ calculatations|
|Generated per visual at runtime|Generated once at import time|

### Removing Bi-directional Filtering Using Filter Measure
**Example:** Fact table **MyFactTable** is filterred by 2 dimensional tables: **Year** + **Geography**
1. Create new measure in Fact table: **_Slicer_MyFactTable_**
    - `Slicer_MyFactTable = INT(NOT ISEMPTY('MyFactTable'))`
    - Returns 1 if atleast one value in the fact table
    - Returns 0 if no values in the fact table
2. Add a visual filter to the slicer and set where `Slicer_MyFactTable = 1`

### Displaying Visuals
- Use restrictive filters to minimize data
- Show as lkittle data as possible on visuals
- Limit the numer of visuals on report pages
- Use only fast customer visuals

# DAX
- DAX is a formula expression language used in multiple Microsoft analytics tools
- DAX formulas include funcstions, operators and values to perform advanced calculations
- DAX formulas are used in:
  - Measures
  - Calculated columns
  - Calculated tables
  - Row-level security

## Calcualted Columns vs. Measures
|Calculated Columns|Measures|
|-|-|
|Calculated on data import|Calculated at query run-time|
|Visible in Table and Report view|Visible only in report pane|
|Example: `COST = Orders[Sales] - Orders[Profit]`|Example: `Total Sales = SUM(Orders[Sales])`|

## DAX Context

### 1. Row Context
- The **_current row_**
- DAX **_calculated columns_**
![](./images/Datacamp/DAX_Row_Context_Example.png)

- Usage:
  - Create new table and columns associated to that table, e.g. `CALENDAR(<start_date>, <end_date>)`
  - Create calculated columns, e.g. Profit = Revenue - Cost
  - Add data cleaning, e.g. `SUBSTITUTE(<text>, <old_text>, <new_text>, <instance_num>)`
  - Add columns from other related table, e.g. adding 1 column from Dim table to Fact table with `RELATED(<column>)`

### 2. Query Context
- Refers to **_subset of data_** that is implicitly retrieved for a formula
- Controlled by slicers, page filters, table columns and row headers
- Controlled by chart/visual filters
- Applies after **_Row Context_**
![](./images/Datacamp/DAX_Query_Context_Example.png)

### 3. Filter Context
- The set of values allowed in each column, or in the values retrieved from a related table
- By using **_arguments to a formula_** or by using report filters on row and column headings
- Applied after **_Query Context_**
![](./images/Datacamp/DAX_Filter_Context_Example.png)
